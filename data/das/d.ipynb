```python
from collections import Counter
class Solution(object):
    def countCharacters(self, words, chars):
        """
        :type words: List[str]
        :type chars: str
        :rtype: int
        """
        total=0
        wo=Counter(chars)
        #print(wo)
        for word in words:
            word_count=Counter(word)
            if all(word_count[ch]<=wo[ch] for ch in word_count):
                total+=len(word)
        return total

        

os=Solution()
words = ["cat","bt","hat","tree"] 
chars = "atach"
print(os.countCharacters(words,chars))
```
